# Ollama (local models)

LiteLLM talks to Ollama the same way it talks to cloud providers.

Ollama must be running locally (`ollama serve`). Pull a model first, for example:

```bash
ollama pull llama3.2
```

````python
from llm_playground.env import load_env
from litellm import completion

load_env()
````

`OLLAMA_API_BASE` defaults to `http://localhost:11434` in `.env.example`.
Change the `ollama/...` model name to whatever you have pulled.

````python
response = completion(
    model="ollama/llama3.2",
    messages=[{"role": "user", "content": "Say hello in one sentence."}],
)
print(response.choices[0].message.content)
````
