# LiteLLM chat

Call cloud models through LiteLLM's unified `completion` API.

Copy `.env.example` to `.env` and set the key for the provider you want to use.

```python
from llm_playground.env import load_env
from litellm import completion

load_env()
```

Switch providers by changing `model`. Examples:

- `gpt-4o-mini` (OpenAI, needs `OPENAI_API_KEY`)
- `anthropic/claude-sonnet-4-5` (needs `ANTHROPIC_API_KEY`)
- `gemini/gemini-2.5-flash` (needs `GEMINI_API_KEY`)

```python
response = completion(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Say hello in one sentence."}],
)
print(response.choices[0].message.content)
```
